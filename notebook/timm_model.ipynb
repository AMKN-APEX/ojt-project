# TIMM モデルの詳細

```python
import timm
import tqdm
from torchinfo import summary
import torch
```

```python
# --- timmモデルの一覧確認 ---
pretrained = False
timm.list_models('resnet*', pretrained = pretrained)
```

```python
# --- 特定のモデルの詳細情報 ---
model_name = 'resnet10t'
x = torch.randn(1, 3, 224, 224)

model = timm.create_model(model_name)
summary(model, input_data=x, col_names=["input_size", "output_size", "num_params", "trainable"])
```
